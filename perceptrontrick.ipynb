```python
from sklearn.datasets import make_classification
import numpy as np
features, labels = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,class_sep=10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.scatter(features[:,0],features[:,1],c=labels,cmap='winter',s=100)
```

```python
def perceptron(features,labels):

    features = np.insert(features,0,1,axis=1)
    weights = np.ones(features.shape[1])
    lr = 0.1

    for i in range(1000):
        j = np.random.randint(0,100)
        labels_hat = step(np.dot(features[j],weights))
        weights = weights + lr*(labels[j]-labels_hat)*features[j]

    return weights[0],weights[1:]
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
intercept_,coef_ = perceptron(features,labels)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
labels_input = m*x_input + b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,labels_input,color='red',linewidth=3)
plt.scatter(features[:,0],features[:,1],c=labels,cmap='winter',s=100)
plt.ylim(-3,2)
```

```python
def perceptron(features,labels):

    m = []
    b = []

    features = np.insert(features,0,1,axis=1)
    weights = np.ones(features.shape[1])
    lr = 0.1

    for i in range(200):
        j = np.random.randint(0,100)
        labels_hat = step(np.dot(features[j],weights))
        weights = weights + lr*(labels[j]-labels_hat)*features[j]

        m.append(-(weights[1]/weights[2]))
        b.append(-(weights[0]/weights[2]))

    return m,b
```

```python
m,b = perceptron(features,labels)
```
